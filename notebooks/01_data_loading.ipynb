```python
# ==========================================================
# PROJECT PHASE 2 - DATA CLEANING
# Question 9
# Identify outlier prices
# ==========================================================

merged_df["original_price_inr"].describe()
```

```python
# ==========================================================
# Find Q1 and Q3
# ==========================================================

Q1 = merged_df["original_price_inr"].quantile(0.25)
Q3 = merged_df["original_price_inr"].quantile(0.75)

print("Q1 =", Q1)
print("Q3 =", Q3)
```

```python
# ==========================================================
# Calculate IQR
# ==========================================================

IQR = Q3 - Q1

print("IQR =", IQR)
```

```python
# ==========================================================
# Find lower and upper limits
# ==========================================================

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

print("Lower Limit =", lower_limit)
print("Upper Limit =", upper_limit)
```

```python
# ==========================================================
# Count outliers
# ==========================================================

outliers = merged_df[
    (merged_df["original_price_inr"] < lower_limit) |
    (merged_df["original_price_inr"] > upper_limit)
]

len(outliers)
```

```python
# ==========================================================
# Replace lower outliers
# ==========================================================

merged_df.loc[
    merged_df["original_price_inr"] < lower_limit,
    "original_price_inr"
] = lower_limit
```

```python
# ==========================================================
# Replace upper outliers
# ==========================================================

merged_df.loc[
    merged_df["original_price_inr"] > upper_limit,
    "original_price_inr"
] = upper_limit
```
